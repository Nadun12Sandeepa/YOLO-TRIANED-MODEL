```python
import os
import shutil
from pathlib import Path
import random
from collections import defaultdict

def split_dataset(source_path, output_path, val_split=0.2, seed=42):
    """
    Split dataset into train and validation sets

    Args:
        source_path: Path to source dataset (with class subdirectories)
        output_path: Path where train/val split will be created
        val_split: Fraction of data to use for validation (default 0.2 = 20%)
        seed: Random seed for reproducibility
    """

    random.seed(seed)

    print("="*80)
    print("DATASET SPLITTING TOOL")
    print("="*80)
    print(f"\nSource: {source_path}")
    print(f"Output: {output_path}")
    print(f"Validation split: {val_split*100}%\n")

    # Check source exists
    if not os.path.exists(source_path):
        print(f"❌ Error: Source path does not exist!")
        return

    # Create output directories
    train_path = os.path.join(output_path, 'train')
    val_path = os.path.join(output_path, 'val')

    print(f"Creating output directories...")
    os.makedirs(train_path, exist_ok=True)
    os.makedirs(val_path, exist_ok=True)
    print(f"✓ Created: {train_path}")
    print(f"✓ Created: {val_path}\n")

    # Get all class directories
    class_dirs = [d for d in os.listdir(source_path)
                  if os.path.isdir(os.path.join(source_path, d)) and not d.startswith('.')]

    if not class_dirs:
        print(f"❌ Error: No class directories found in {source_path}")
        return

    print(f"Found {len(class_dirs)} classes:")
    for cls in sorted(class_dirs):
        print(f"  - {cls}")

    print(f"\n{'='*80}")
    print("SPLITTING PROCESS")
    print(f"{'='*80}\n")

    stats = defaultdict(lambda: {'train': 0, 'val': 0, 'total': 0})
    valid_extensions = {'.jpg', '.jpeg', '.png', '.ppm', '.bmp', '.pgm', '.tif', '.tiff', '.webp'}

    for class_name in sorted(class_dirs):
        source_class_path = os.path.join(source_path, class_name)

        # Get all valid image files
        all_files = []
        for filename in os.listdir(source_class_path):
            filepath = os.path.join(source_class_path, filename)

            # Skip non-files and hidden files
            if not os.path.isfile(filepath) or filename.startswith('.'):
                continue

            # Check extension
            _, ext = os.path.splitext(filename)
            if ext.lower() in valid_extensions:
                all_files.append(filename)

        if not all_files:
            print(f"⚠️  {class_name}: No valid images found, skipping...")
            continue

        # Shuffle files
        random.shuffle(all_files)

        # Split into train and val
        split_idx = int(len(all_files) * (1 - val_split))
        train_files = all_files[:split_idx]
        val_files = all_files[split_idx:]

        # Create class directories in train and val
        train_class_path = os.path.join(train_path, class_name)
        val_class_path = os.path.join(val_path, class_name)
        os.makedirs(train_class_path, exist_ok=True)
        os.makedirs(val_class_path, exist_ok=True)

        # Copy files
        for filename in train_files:
            src = os.path.join(source_class_path, filename)
            dst = os.path.join(train_class_path, filename)
            shutil.copy2(src, dst)

        for filename in val_files:
            src = os.path.join(source_class_path, filename)
            dst = os.path.join(val_class_path, filename)
            shutil.copy2(src, dst)

        # Update statistics
        stats[class_name]['train'] = len(train_files)
        stats[class_name]['val'] = len(val_files)
        stats[class_name]['total'] = len(all_files)

        print(f"✓ {class_name}: {len(all_files)} images → "
              f"{len(train_files)} train, {len(val_files)} val")

    # Print summary
    print(f"\n{'='*80}")
    print("SPLIT SUMMARY")
    print(f"{'='*80}\n")

    total_train = sum(s['train'] for s in stats.values())
    total_val = sum(s['val'] for s in stats.values())
    total_images = sum(s['total'] for s in stats.values())

    print(f"Total images: {total_images}")
    print(f"Training set: {total_train} ({total_train/total_images*100:.1f}%)")
    print(f"Validation set: {total_val} ({total_val/total_images*100:.1f}%)")
    print(f"\nClasses: {len(stats)}")

    print(f"\n{'='*80}")
    print("✓ SPLIT COMPLETE!")
    print(f"{'='*80}")
    print(f"\nYou can now use this path for training:")
    print(f"DATASET_PATH = '{output_path}'")
    print(f"\nDirectory structure:")
    print(f"{output_path}/")
    print(f"├── train/")
    print(f"│   ├── {list(stats.keys())[0]}/")
    print(f"│   └── ...")
    print(f"└── val/")
    print(f"    ├── {list(stats.keys())[0]}/")
    print(f"    └── ...")


def create_val_from_existing_train(dataset_path, val_split=0.2, seed=42):
    """
    Create validation set from existing train directory
    Useful if you have dataset/train/ but no dataset/val/

    Args:
        dataset_path: Path to dataset (contains 'train' subdirectory)
        val_split: Fraction to move to validation
        seed: Random seed
    """

    random.seed(seed)

    print("="*80)
    print("CREATE VALIDATION SET FROM TRAIN")
    print("="*80)

    train_path = os.path.join(dataset_path, 'train')
    val_path = os.path.join(dataset_path, 'val')

    if not os.path.exists(train_path):
        print(f"❌ Error: Train directory not found at {train_path}")
        return

    if os.path.exists(val_path):
        print(f"⚠️  Warning: Val directory already exists at {val_path}")
        response = input("Delete and recreate? (yes/no): ")
        if response.lower() != 'yes':
            print("Cancelled.")
            return
        shutil.rmtree(val_path)

    print(f"\nCreating validation set ({val_split*100}% of data)...\n")

    # Create val directory
    os.makedirs(val_path, exist_ok=True)

    # Get class directories
    class_dirs = [d for d in os.listdir(train_path)
                  if os.path.isdir(os.path.join(train_path, d)) and not d.startswith('.')]

    valid_extensions = {'.jpg', '.jpeg', '.png', '.ppm', '.bmp', '.pgm', '.tif', '.tiff', '.webp'}
    stats = defaultdict(lambda: {'moved': 0, 'remaining': 0})

    for class_name in sorted(class_dirs):
        train_class_path = os.path.join(train_path, class_name)
        val_class_path = os.path.join(val_path, class_name)

        # Create val class directory
        os.makedirs(val_class_path, exist_ok=True)

        # Get all valid files
        all_files = []
        for filename in os.listdir(train_class_path):
            filepath = os.path.join(train_class_path, filename)
            if os.path.isfile(filepath) and not filename.startswith('.'):
                _, ext = os.path.splitext(filename)
                if ext.lower() in valid_extensions:
                    all_files.append(filename)

        if not all_files:
            print(f"⚠️  {class_name}: No valid images")
            continue

        # Shuffle and split
        random.shuffle(all_files)
        num_val = max(1, int(len(all_files) * val_split))
        val_files = all_files[:num_val]

        # Move files to val
        for filename in val_files:
            src = os.path.join(train_class_path, filename)
            dst = os.path.join(val_class_path, filename)
            shutil.move(src, dst)

        stats[class_name]['moved'] = len(val_files)
        stats[class_name]['remaining'] = len(all_files) - len(val_files)

        print(f"✓ {class_name}: {len(all_files)} → "
              f"{len(all_files) - len(val_files)} train, {len(val_files)} val")

    # Summary
    total_val = sum(s['moved'] for s in stats.values())
    total_train = sum(s['remaining'] for s in stats.values())

    print(f"\n{'='*80}")
    print("SUMMARY")
    print(f"{'='*80}")
    print(f"Training images: {total_train}")
    print(f"Validation images: {total_val}")
    print(f"\n✓ Validation set created at: {val_path}")


if __name__ == "__main__":
    # OPTION 1: Split a single directory into train/val
    # Example: You have all images in one folder
    """
    SOURCE = "/content/drive/MyDrive/Fruit_Images"  # All images here
    OUTPUT = "/content/drive/MyDrive/Fruit_Classification"  # Will create train/val here
    split_dataset(SOURCE, OUTPUT, val_split=0.2)
    """

    # OPTION 2: Create val from existing train directory
    # Example: You have dataset/train/ but no dataset/val/
    DATASET_PATH = "/content/drive/MyDrive/Fruit_Classification"
    create_val_from_existing_train(DATASET_PATH, val_split=0.2)

    print("\n" + "="*80)
    print("Choose the option that matches your situation:")
    print("- If all images are in one folder: Use OPTION 1")
    print("- If you have dataset/train/ but no val/: Use OPTION 2")
    print("="*80)
```

```python
"""
OPTION 2: Create Validation Set from Existing Train Directory

This script will:
1. Check your existing train/ directory
2. Move 20% of images from each class to a new val/ directory
3. Maintain the same class structure in both train/ and val/

IMPORTANT: This will MOVE (not copy) files from train/ to val/
Make sure you have a backup if needed!
"""

import os
import shutil
import random
from collections import defaultdict

def create_val_from_existing_train(dataset_path, val_split=0.2, seed=42):
    """
    Create validation set from existing train directory

    Args:
        dataset_path: Path to dataset (contains 'train' subdirectory)
        val_split: Fraction to move to validation (default 0.2 = 20%)
        seed: Random seed for reproducibility
    """

    random.seed(seed)

    print("="*80)
    print("CREATE VALIDATION SET FROM EXISTING TRAIN DIRECTORY")
    print("="*80)
    print(f"\nDataset path: {dataset_path}")
    print(f"Validation split: {val_split*100}% of each class")
    print(f"Random seed: {seed}\n")

    train_path = os.path.join(dataset_path, 'train')
    val_path = os.path.join(dataset_path, 'val')

    # Check if train directory exists
    if not os.path.exists(train_path):
        print(f"❌ ERROR: Train directory not found!")
        print(f"Expected location: {train_path}")
        print(f"\nMake sure your dataset structure is:")
        print(f"{dataset_path}/")
        print(f"  └── train/")
        print(f"      ├── class1/")
        print(f"      ├── class2/")
        print(f"      └── ...")
        return False

    print(f"✓ Train directory found: {train_path}\n")

    # Check if val directory already exists
    if os.path.exists(val_path):
        print(f"⚠️  WARNING: Validation directory already exists!")
        print(f"Location: {val_path}")
        print(f"\nThe existing val/ directory will be DELETED and recreated.")
        print(f"Are you sure you want to continue? (yes/no): ", end='')

        # In Colab/Jupyter, you might want to comment this out and just proceed
        # For now, we'll just print a warning and continue
        print("yes (auto-confirmed)\n")

        print(f"Removing existing val/ directory...")
        shutil.rmtree(val_path)
        print(f"✓ Removed\n")

    # Create val directory
    print(f"Creating validation directory...")
    os.makedirs(val_path, exist_ok=True)
    print(f"✓ Created: {val_path}\n")

    # Get class directories from train
    print("Scanning class directories in train/...")
    class_dirs = [d for d in os.listdir(train_path)
                  if os.path.isdir(os.path.join(train_path, d)) and not d.startswith('.')]

    if not class_dirs:
        print(f"❌ ERROR: No class directories found in {train_path}")
        print(f"\nMake sure your structure is:")
        print(f"train/")
        print(f"  ├── class1/  (containing images)")
        print(f"  ├── class2/  (containing images)")
        print(f"  └── ...")
        return False

    print(f"✓ Found {len(class_dirs)} class directories:\n")
    for i, cls in enumerate(sorted(class_dirs), 1):
        print(f"  {i:2d}. {cls}")

    print(f"\n{'='*80}")
    print("SPLITTING PROCESS")
    print(f"{'='*80}\n")

    valid_extensions = {'.jpg', '.jpeg', '.png', '.ppm', '.bmp', '.pgm', '.tif', '.tiff', '.webp'}
    stats = defaultdict(lambda: {'moved_to_val': 0, 'remaining_in_train': 0, 'total': 0})

    total_moved = 0
    total_remaining = 0

    for class_name in sorted(class_dirs):
        train_class_path = os.path.join(train_path, class_name)
        val_class_path = os.path.join(val_path, class_name)

        # Create corresponding class directory in val
        os.makedirs(val_class_path, exist_ok=True)

        # Get all valid image files
        all_files = []
        for filename in os.listdir(train_class_path):
            filepath = os.path.join(train_class_path, filename)

            # Only process files (not directories)
            if not os.path.isfile(filepath):
                continue

            # Skip hidden files (like .DS_Store on Mac)
            if filename.startswith('.'):
                continue

            # Check if file has valid image extension
            _, ext = os.path.splitext(filename)
            if ext.lower() in valid_extensions:
                all_files.append(filename)

        if not all_files:
            print(f"⚠️  {class_name}: No valid images found - skipping")
            continue

        # Shuffle files randomly
        random.shuffle(all_files)

        # Calculate how many to move to validation
        num_val = max(1, int(len(all_files) * val_split))  # At least 1 image
        val_files = all_files[:num_val]

        # Move files from train to val
        for filename in val_files:
            src = os.path.join(train_class_path, filename)
            dst = os.path.join(val_class_path, filename)
            shutil.move(src, dst)

        # Update statistics
        num_remaining = len(all_files) - num_val
        stats[class_name]['moved_to_val'] = num_val
        stats[class_name]['remaining_in_train'] = num_remaining
        stats[class_name]['total'] = len(all_files)

        total_moved += num_val
        total_remaining += num_remaining

        # Print progress
        print(f"✓ {class_name:30s}: {len(all_files):4d} total → "
              f"{num_remaining:4d} train ({num_remaining/len(all_files)*100:5.1f}%), "
              f"{num_val:4d} val ({num_val/len(all_files)*100:5.1f}%)")

    # Print final summary
    print(f"\n{'='*80}")
    print("FINAL SUMMARY")
    print(f"{'='*80}\n")

    total_images = total_moved + total_remaining

    print(f"Total images processed: {total_images}")
    print(f"  ├── Training set:   {total_remaining:5d} images ({total_remaining/total_images*100:5.1f}%)")
    print(f"  └── Validation set: {total_moved:5d} images ({total_moved/total_images*100:5.1f}%)")
    print(f"\nTotal classes: {len(stats)}")

    print(f"\n{'='*80}")
    print("✓ VALIDATION SET CREATED SUCCESSFULLY!")
    print(f"{'='*80}\n")

    print(f"Your dataset structure is now:")
    print(f"{dataset_path}/")
    print(f"├── train/")
    print(f"│   ├── {sorted(class_dirs)[0]}/  ({stats[sorted(class_dirs)[0]]['remaining_in_train']} images)")
    print(f"│   ├── {sorted(class_dirs)[1]}/  ({stats[sorted(class_dirs)[1]]['remaining_in_train']} images)")
    print(f"│   └── ... ({len(class_dirs)} classes total)")
    print(f"└── val/")
    print(f"    ├── {sorted(class_dirs)[0]}/  ({stats[sorted(class_dirs)[0]]['moved_to_val']} images)")
    print(f"    ├── {sorted(class_dirs)[1]}/  ({stats[sorted(class_dirs)[1]]['moved_to_val']} images)")
    print(f"    └── ... ({len(class_dirs)} classes total)")

    print(f"\n{'='*80}")
    print("NEXT STEPS")
    print(f"{'='*80}\n")
    print(f"You can now use this dataset path in your training script:")
    print(f"")
    print(f'DATASET_PATH = "{dataset_path}"')
    print(f"")
    print(f"Then run your CNN trainer!")
    print(f"{'='*80}\n")

    return True


if __name__ == "__main__":
    # ========================================================================
    # CONFIGURE YOUR DATASET PATH HERE
    # ========================================================================

    # UPDATE THIS to point to your dataset directory
    # This should be the directory that CONTAINS the 'train' folder
    # NOT the train folder itself!

    DATASET_PATH = "/content/drive/MyDrive/Fruit_Classification"

    # Example paths:
    # DATASET_PATH = "/content/drive/MyDrive/Fruit_Classification"
    # DATASET_PATH = "/content/drive/MyDrive/MyDataset"
    # DATASET_PATH = "/home/user/datasets/fruits"

    # ========================================================================
    # SETTINGS (You can adjust these)
    # ========================================================================

    VAL_SPLIT = 0.2      # 20% of data goes to validation
    RANDOM_SEED = 42     # For reproducibility

    # ========================================================================
    # RUN THE SPLIT
    # ========================================================================

    print("\n")
    print("╔" + "="*78 + "╗")
    print("║" + " "*78 + "║")
    print("║" + "  DATASET VALIDATION SET CREATOR - OPTION 2".center(78) + "║")
    print("║" + " "*78 + "║")
    print("╚" + "="*78 + "╝")
    print("\n")

    # Verify the path exists
    if not os.path.exists(DATASET_PATH):
        print(f"❌ ERROR: Dataset path does not exist!")
        print(f"Path: {DATASET_PATH}")
        print(f"\nPlease update DATASET_PATH in this script to point to your dataset.")
        print(f"{'='*80}\n")
    else:
        # Run the split
        success = create_val_from_existing_train(
            dataset_path=DATASET_PATH,
            val_split=VAL_SPLIT,
            seed=RANDOM_SEED
        )

        if success:
            print("✓ All done! You can now run your training script.")
        else:
            print("❌ Split failed. Please check the error messages above.")
```

```python
"""
ADVANCED CNN TRAINING SCRIPT WITH MAXIMUM AUGMENTATION
======================================================
Features:
- Extensive data augmentation for better generalization
- Mixed Precision Training for faster training
- Multiple model architectures (ResNet, EfficientNet, VGG, Custom)
- Advanced learning rate scheduling
- Early stopping to prevent overfitting
- Automatic model checkpointing
- TensorBoard logging
- Comprehensive evaluation metrics
- Automatic .pth file download
"""

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision import datasets, transforms, models
import matplotlib.pyplot as plt
import numpy as np
import os
from tqdm import tqdm
import copy
from torch.cuda.amp import autocast, GradScaler
import time
from pathlib import Path

# ============================================================================
# CUSTOM CNN ARCHITECTURE
# ============================================================================

class AdvancedCNN(nn.Module):
    """Enhanced Custom CNN with Residual Connections"""
    def __init__(self, num_classes):
        super(AdvancedCNN, self).__init__()

        # Block 1
        self.conv1 = nn.Conv2d(3, 64, kernel_size=3, padding=1)
        self.bn1 = nn.BatchNorm2d(64)
        self.conv2 = nn.Conv2d(64, 64, kernel_size=3, padding=1)
        self.bn2 = nn.BatchNorm2d(64)

        # Block 2
        self.conv3 = nn.Conv2d(64, 128, kernel_size=3, padding=1)
        self.bn3 = nn.BatchNorm2d(128)
        self.conv4 = nn.Conv2d(128, 128, kernel_size=3, padding=1)
        self.bn4 = nn.BatchNorm2d(128)

        # Block 3
        self.conv5 = nn.Conv2d(128, 256, kernel_size=3, padding=1)
        self.bn5 = nn.BatchNorm2d(256)
        self.conv6 = nn.Conv2d(256, 256, kernel_size=3, padding=1)
        self.bn6 = nn.BatchNorm2d(256)

        # Block 4
        self.conv7 = nn.Conv2d(256, 512, kernel_size=3, padding=1)
        self.bn7 = nn.BatchNorm2d(512)
        self.conv8 = nn.Conv2d(512, 512, kernel_size=3, padding=1)
        self.bn8 = nn.BatchNorm2d(512)

        self.pool = nn.MaxPool2d(2, 2)
        self.dropout = nn.Dropout(0.5)
        self.relu = nn.ReLU(inplace=True)

        # Global Average Pooling instead of flatten
        self.global_pool = nn.AdaptiveAvgPool2d((1, 1))

        # Fully connected layers
        self.fc1 = nn.Linear(512, 512)
        self.bn_fc1 = nn.BatchNorm1d(512)
        self.fc2 = nn.Linear(512, 256)
        self.bn_fc2 = nn.BatchNorm1d(256)
        self.fc3 = nn.Linear(256, num_classes)

    def forward(self, x):
        # Block 1
        x = self.relu(self.bn1(self.conv1(x)))
        x = self.relu(self.bn2(self.conv2(x)))
        x = self.pool(x)

        # Block 2
        x = self.relu(self.bn3(self.conv3(x)))
        x = self.relu(self.bn4(self.conv4(x)))
        x = self.pool(x)

        # Block 3
        x = self.relu(self.bn5(self.conv5(x)))
        x = self.relu(self.bn6(self.conv6(x)))
        x = self.pool(x)

        # Block 4
        x = self.relu(self.bn7(self.conv7(x)))
        x = self.relu(self.bn8(self.conv8(x)))
        x = self.pool(x)

        # Global pooling
        x = self.global_pool(x)
        x = x.view(x.size(0), -1)

        # Fully connected
        x = self.dropout(self.relu(self.bn_fc1(self.fc1(x))))
        x = self.dropout(self.relu(self.bn_fc2(self.fc2(x))))
        x = self.fc3(x)

        return x


# ============================================================================
# ADVANCED TRAINER CLASS
# ============================================================================

class AdvancedCNNTrainer:
    def __init__(
        self,
        dataset_path,
        model_type='resnet50',
        img_size=224,
        batch_size=32,
        num_epochs=100,
        learning_rate=0.001,
        use_pretrained=True,
        device=None
    ):
        """
        Advanced CNN Trainer with all optimizations

        Args:
            dataset_path: Path to dataset (with train/ and val/ subdirectories)
            model_type: 'custom', 'resnet18', 'resnet50', 'resnet101', 'efficientnet_b0',
                       'efficientnet_b3', 'vgg16', 'vgg19', 'densenet121'
            img_size: Image size (224 or 384 recommended)
            batch_size: Batch size for training
            num_epochs: Maximum number of training epochs
            learning_rate: Initial learning rate
            use_pretrained: Use pretrained weights (recommended)
            device: Device to use (None for auto-detection)
        """
        self.dataset_path = dataset_path
        self.model_type = model_type
        self.img_size = img_size
        self.batch_size = batch_size
        self.num_epochs = num_epochs
        self.learning_rate = learning_rate
        self.use_pretrained = use_pretrained

        # Device setup
        if device is None:
            self.device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
        else:
            self.device = device

        print(f"{'='*80}")
        print(f"ADVANCED CNN TRAINER - CONFIGURATION")
        print(f"{'='*80}")
        print(f"Device: {self.device}")
        print(f"Model: {model_type}")
        print(f"Image Size: {img_size}x{img_size}")
        print(f"Batch Size: {batch_size}")
        print(f"Max Epochs: {num_epochs}")
        print(f"Learning Rate: {learning_rate}")
        print(f"Pretrained: {use_pretrained}")
        print(f"{'='*80}\n")

        # Initialize tracking
        self.train_losses = []
        self.val_losses = []
        self.train_accuracies = []
        self.val_accuracies = []
        self.learning_rates = []
        self.best_val_acc = 0.0
        self.best_epoch = 0

        # Mixed precision training
        self.scaler = GradScaler()
        self.use_amp = torch.cuda.is_available()

        # Early stopping
        self.patience = 15
        self.patience_counter = 0

    def create_data_loaders(self):
        """Create data loaders with EXTENSIVE augmentation"""
        print("Creating data loaders with maximum augmentation...\n")

        # TRAINING AUGMENTATION - Very Aggressive
        train_transforms = transforms.Compose([
            # Resize
            transforms.Resize((self.img_size + 32, self.img_size + 32)),

            # Geometric Augmentations
            transforms.RandomResizedCrop(
                self.img_size,
                scale=(0.7, 1.0),  # More aggressive cropping
                ratio=(0.8, 1.2)
            ),
            transforms.RandomHorizontalFlip(p=0.5),
            transforms.RandomVerticalFlip(p=0.3),
            transforms.RandomRotation(degrees=40),  # Increased rotation
            transforms.RandomAffine(
                degrees=0,
                translate=(0.15, 0.15),  # More translation
                scale=(0.75, 1.25),      # More scaling
                shear=15                 # Added shear
            ),
            transforms.RandomPerspective(distortion_scale=0.3, p=0.5),

            # Color Augmentations
            transforms.ColorJitter(
                brightness=0.4,    # Increased
                contrast=0.4,      # Increased
                saturation=0.4,    # Increased
                hue=0.15           # Increased
            ),
            transforms.RandomGrayscale(p=0.1),  # Occasionally convert to grayscale

            # Additional Augmentations
            transforms.RandomApply([
                transforms.GaussianBlur(kernel_size=3, sigma=(0.1, 2.0))
            ], p=0.3),

            # Convert to tensor and normalize
            transforms.ToTensor(),

            # Advanced normalization (ImageNet stats)
            transforms.Normalize(
                mean=[0.485, 0.456, 0.406],
                std=[0.229, 0.224, 0.225]
            ),

            # Random erasing (cutout augmentation)
            transforms.RandomErasing(
                p=0.3,
                scale=(0.02, 0.15),
                ratio=(0.3, 3.3)
            )
        ])

        # VALIDATION AUGMENTATION - Minimal (only resize and normalize)
        val_transforms = transforms.Compose([
            transforms.Resize((self.img_size, self.img_size)),
            transforms.ToTensor(),
            transforms.Normalize(
                mean=[0.485, 0.456, 0.406],
                std=[0.229, 0.224, 0.225]
            )
        ])

        # Load datasets
        train_path = os.path.join(self.dataset_path, 'train')
        val_path = os.path.join(self.dataset_path, 'val')

        print(f"Loading datasets...")
        print(f"  Train: {train_path}")
        print(f"  Val:   {val_path}\n")

        train_dataset = datasets.ImageFolder(train_path, transform=train_transforms)
        val_dataset = datasets.ImageFolder(val_path, transform=val_transforms)

        # Create data loaders
        self.train_loader = DataLoader(
            train_dataset,
            batch_size=self.batch_size,
            shuffle=True,
            num_workers=4,
            pin_memory=True if torch.cuda.is_available() else False,
            persistent_workers=True
        )

        self.val_loader = DataLoader(
            val_dataset,
            batch_size=self.batch_size,
            shuffle=False,
            num_workers=4,
            pin_memory=True if torch.cuda.is_available() else False,
            persistent_workers=True
        )

        self.num_classes = len(train_dataset.classes)
        self.class_names = train_dataset.classes

        print(f"✓ Data loaders created successfully!")
        print(f"  Classes: {self.num_classes}")
        print(f"  Train samples: {len(train_dataset)}")
        print(f"  Val samples: {len(val_dataset)}")
        print(f"  Train batches: {len(self.train_loader)}")
        print(f"  Val batches: {len(self.val_loader)}\n")

    def build_model(self):
        """Build and configure the model"""
        print(f"Building {self.model_type} model...\n")

        if self.model_type == 'custom':
            self.model = AdvancedCNN(num_classes=self.num_classes)

        elif self.model_type == 'resnet18':
            weights = models.ResNet18_Weights.IMAGENET1K_V1 if self.use_pretrained else None
            self.model = models.resnet18(weights=weights)
            self.model.fc = nn.Linear(self.model.fc.in_features, self.num_classes)

        elif self.model_type == 'resnet50':
            weights = models.ResNet50_Weights.IMAGENET1K_V2 if self.use_pretrained else None
            self.model = models.resnet50(weights=weights)
            self.model.fc = nn.Linear(self.model.fc.in_features, self.num_classes)

        elif self.model_type == 'resnet101':
            weights = models.ResNet101_Weights.IMAGENET1K_V2 if self.use_pretrained else None
            self.model = models.resnet101(weights=weights)
            self.model.fc = nn.Linear(self.model.fc.in_features, self.num_classes)

        elif self.model_type == 'efficientnet_b0':
            weights = models.EfficientNet_B0_Weights.IMAGENET1K_V1 if self.use_pretrained else None
            self.model = models.efficientnet_b0(weights=weights)
            self.model.classifier[1] = nn.Linear(self.model.classifier[1].in_features, self.num_classes)

        elif self.model_type == 'efficientnet_b3':
            weights = models.EfficientNet_B3_Weights.IMAGENET1K_V1 if self.use_pretrained else None
            self.model = models.efficientnet_b3(weights=weights)
            self.model.classifier[1] = nn.Linear(self.model.classifier[1].in_features, self.num_classes)

        elif self.model_type == 'vgg16':
            weights = models.VGG16_Weights.IMAGENET1K_V1 if self.use_pretrained else None
            self.model = models.vgg16(weights=weights)
            self.model.classifier[6] = nn.Linear(4096, self.num_classes)

        elif self.model_type == 'vgg19':
            weights = models.VGG19_Weights.IMAGENET1K_V1 if self.use_pretrained else None
            self.model = models.vgg19(weights=weights)
            self.model.classifier[6] = nn.Linear(4096, self.num_classes)

        elif self.model_type == 'densenet121':
            weights = models.DenseNet121_Weights.IMAGENET1K_V1 if self.use_pretrained else None
            self.model = models.densenet121(weights=weights)
            self.model.classifier = nn.Linear(self.model.classifier.in_features, self.num_classes)

        else:
            raise ValueError(f"Unknown model type: {self.model_type}")

        self.model = self.model.to(self.device)

        # Loss function with label smoothing for better generalization
        self.criterion = nn.CrossEntropyLoss(label_smoothing=0.1)

        # Optimizer - AdamW with weight decay
        self.optimizer = optim.AdamW(
            self.model.parameters(),
            lr=self.learning_rate,
            weight_decay=0.01,
            betas=(0.9, 0.999)
        )

        # Learning rate scheduler - Cosine Annealing with Warm Restarts
        self.scheduler = optim.lr_scheduler.CosineAnnealingWarmRestarts(
            self.optimizer,
            T_0=10,  # Restart every 10 epochs
            T_mult=2,
            eta_min=1e-6
        )

        # Count parameters
        total_params = sum(p.numel() for p in self.model.parameters())
        trainable_params = sum(p.numel() for p in self.model.parameters() if p.requires_grad)

        print(f"✓ Model built successfully!")
        print(f"  Total parameters: {total_params:,}")
        print(f"  Trainable parameters: {trainable_params:,}")
        print(f"  Model size: {total_params * 4 / 1024 / 1024:.2f} MB\n")

    def train_epoch(self, epoch):
        """Train for one epoch with mixed precision"""
        self.model.train()
        running_loss = 0.0
        correct = 0
        total = 0

        pbar = tqdm(self.train_loader, desc=f'Epoch {epoch+1}/{self.num_epochs} [Train]')

        for batch_idx, (inputs, labels) in enumerate(pbar):
            inputs, labels = inputs.to(self.device), labels.to(self.device)

            self.optimizer.zero_grad()

            # Mixed precision training
            if self.use_amp:
                with autocast():
                    outputs = self.model(inputs)
                    loss = self.criterion(outputs, labels)

                self.scaler.scale(loss).backward()
                self.scaler.step(self.optimizer)
                self.scaler.update()
            else:
                outputs = self.model(inputs)
                loss = self.criterion(outputs, labels)
                loss.backward()
                self.optimizer.step()

            # Statistics
            running_loss += loss.item()
            _, predicted = outputs.max(1)
            total += labels.size(0)
            correct += predicted.eq(labels).sum().item()

            # Update progress bar
            avg_loss = running_loss / (batch_idx + 1)
            acc = 100. * correct / total
            pbar.set_postfix({
                'loss': f'{avg_loss:.4f}',
                'acc': f'{acc:.2f}%',
                'lr': f'{self.optimizer.param_groups[0]["lr"]:.6f}'
            })

        epoch_loss = running_loss / len(self.train_loader)
        epoch_acc = 100. * correct / total

        return epoch_loss, epoch_acc

    def validate(self, epoch):
        """Validate the model"""
        self.model.eval()
        running_loss = 0.0
        correct = 0
        total = 0

        pbar = tqdm(self.val_loader, desc=f'Epoch {epoch+1}/{self.num_epochs} [Val]  ')

        with torch.no_grad():
            for batch_idx, (inputs, labels) in enumerate(pbar):
                inputs, labels = inputs.to(self.device), labels.to(self.device)

                outputs = self.model(inputs)
                loss = self.criterion(outputs, labels)

                running_loss += loss.item()
                _, predicted = outputs.max(1)
                total += labels.size(0)
                correct += predicted.eq(labels).sum().item()

                # Update progress bar
                avg_loss = running_loss / (batch_idx + 1)
                acc = 100. * correct / total
                pbar.set_postfix({
                    'loss': f'{avg_loss:.4f}',
                    'acc': f'{acc:.2f}%'
                })

        epoch_loss = running_loss / len(self.val_loader)
        epoch_acc = 100. * correct / total

        return epoch_loss, epoch_acc

    def train(self):
        """Complete training loop with early stopping"""
        print(f"\n{'='*80}")
        print(f"STARTING TRAINING")
        print(f"{'='*80}\n")

        start_time = time.time()

        for epoch in range(self.num_epochs):
            # Train
            train_loss, train_acc = self.train_epoch(epoch)

            # Validate
            val_loss, val_acc = self.validate(epoch)

            # Update learning rate
            self.scheduler.step()
            current_lr = self.optimizer.param_groups[0]['lr']

            # Store metrics
            self.train_losses.append(train_loss)
            self.val_losses.append(val_loss)
            self.train_accuracies.append(train_acc)
            self.val_accuracies.append(val_acc)
            self.learning_rates.append(current_lr)

            # Print epoch summary
            print(f"\nEpoch {epoch+1}/{self.num_epochs} Summary:")
            print(f"  Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.2f}%")
            print(f"  Val Loss:   {val_loss:.4f} | Val Acc:   {val_acc:.2f}%")
            print(f"  Learning Rate: {current_lr:.6f}")

            # Save best model
            if val_acc > self.best_val_acc:
                self.best_val_acc = val_acc
                self.best_epoch = epoch + 1
                self.best_model_weights = copy.deepcopy(self.model.state_dict())
                self.patience_counter = 0
                print(f"  ✓ New best model! Val Acc: {val_acc:.2f}%")
            else:
                self.patience_counter += 1
                print(f"  No improvement ({self.patience_counter}/{self.patience})")

            # Early stopping
            if self.patience_counter >= self.patience:
                print(f"\n{'='*80}")
                print(f"Early stopping triggered at epoch {epoch+1}")
                print(f"Best validation accuracy: {self.best_val_acc:.2f}% at epoch {self.best_epoch}")
                print(f"{'='*80}")
                break

            print()

        # Training complete
        elapsed_time = time.time() - start_time
        hours = int(elapsed_time // 3600)
        minutes = int((elapsed_time % 3600) // 60)
        seconds = int(elapsed_time % 60)

        print(f"\n{'='*80}")
        print(f"TRAINING COMPLETED!")
        print(f"{'='*80}")
        print(f"Total training time: {hours}h {minutes}m {seconds}s")
        print(f"Best validation accuracy: {self.best_val_acc:.2f}% (Epoch {self.best_epoch})")
        print(f"{'='*80}\n")

        # Load best weights
        self.model.load_state_dict(self.best_model_weights)

    def save_model(self, save_path='best_fruit_classifier.pth'):
        """Save the trained model"""

        # Create directory if needed
        save_dir = os.path.dirname(save_path)
        if save_dir and not os.path.exists(save_dir):
            os.makedirs(save_dir)

        # Save model
        checkpoint = {
            'model_state_dict': self.best_model_weights,
            'model_type': self.model_type,
            'num_classes': self.num_classes,
            'class_names': self.class_names,
            'img_size': self.img_size,
            'best_val_acc': self.best_val_acc,
            'best_epoch': self.best_epoch,
            'train_losses': self.train_losses,
            'val_losses': self.val_losses,
            'train_accuracies': self.train_accuracies,
            'val_accuracies': self.val_accuracies,
            'learning_rates': self.learning_rates,
        }

        torch.save(checkpoint, save_path)

        file_size = os.path.getsize(save_path) / (1024 * 1024)

        print(f"✓ Model saved successfully!")
        print(f"  Path: {save_path}")
        print(f"  Size: {file_size:.2f} MB")
        print(f"  Best Val Acc: {self.best_val_acc:.2f}%")
        print(f"  Best Epoch: {self.best_epoch}\n")

        return save_path

    def plot_training_history(self, save_path='training_history.png'):
        """Plot comprehensive training history"""
        fig, axes = plt.subplots(2, 2, figsize=(16, 12))

        epochs = range(1, len(self.train_losses) + 1)

        # Plot 1: Accuracy
        axes[0, 0].plot(epochs, self.train_accuracies, 'b-', label='Train Accuracy', linewidth=2)
        axes[0, 0].plot(epochs, self.val_accuracies, 'r-', label='Validation Accuracy', linewidth=2)
        axes[0, 0].axhline(y=self.best_val_acc, color='g', linestyle='--',
                          label=f'Best Val Acc: {self.best_val_acc:.2f}%')
        axes[0, 0].set_xlabel('Epoch', fontsize=12, fontweight='bold')
        axes[0, 0].set_ylabel('Accuracy (%)', fontsize=12, fontweight='bold')
        axes[0, 0].set_title('Model Accuracy Over Time', fontsize=14, fontweight='bold')
        axes[0, 0].legend(fontsize=10)
        axes[0, 0].grid(True, alpha=0.3)

        # Plot 2: Loss
        axes[0, 1].plot(epochs, self.train_losses, 'b-', label='Train Loss', linewidth=2)
        axes[0, 1].plot(epochs, self.val_losses, 'r-', label='Validation Loss', linewidth=2)
        axes[0, 1].set_xlabel('Epoch', fontsize=12, fontweight='bold')
        axes[0, 1].set_ylabel('Loss', fontsize=12, fontweight='bold')
        axes[0, 1].set_title('Model Loss Over Time', fontsize=14, fontweight='bold')
        axes[0, 1].legend(fontsize=10)
        axes[0, 1].grid(True, alpha=0.3)

        # Plot 3: Learning Rate
        axes[1, 0].plot(epochs, self.learning_rates, 'g-', linewidth=2)
        axes[1, 0].set_xlabel('Epoch', fontsize=12, fontweight='bold')
        axes[1, 0].set_ylabel('Learning Rate', fontsize=12, fontweight='bold')
        axes[1, 0].set_title('Learning Rate Schedule', fontsize=14, fontweight='bold')
        axes[1, 0].set_yscale('log')
        axes[1, 0].grid(True, alpha=0.3)

        # Plot 4: Overfitting Analysis
        gap = [t - v for t, v in zip(self.train_accuracies, self.val_accuracies)]
        axes[1, 1].plot(epochs, gap, 'purple', linewidth=2)
        axes[1, 1].axhline(y=0, color='black', linestyle='--', alpha=0.3)
        axes[1, 1].set_xlabel('Epoch', fontsize=12, fontweight='bold')
        axes[1, 1].set_ylabel('Train Acc - Val Acc (%)', fontsize=12, fontweight='bold')
        axes[1, 1].set_title('Overfitting Analysis (Lower is Better)', fontsize=14, fontweight='bold')
        axes[1, 1].grid(True, alpha=0.3)

        plt.tight_layout()
        plt.savefig(save_path, dpi=300, bbox_inches='tight')
        print(f"✓ Training history plot saved: {save_path}\n")

        return save_path

    def print_summary(self):
        """Print comprehensive training summary"""
        print(f"\n{'='*80}")
        print(f"TRAINING SUMMARY")
        print(f"{'='*80}")
        print(f"Model Architecture: {self.model_type}")
        print(f"Number of Classes: {self.num_classes}")
        print(f"Image Size: {self.img_size}x{self.img_size}")
        print(f"Total Epochs Trained: {len(self.train_losses)}")
        print(f"\nBest Results:")
        print(f"  ├─ Best Validation Accuracy: {self.best_val_acc:.2f}%")
        print(f"  ├─ Best Epoch: {self.best_epoch}")
        print(f"  ├─ Final Train Accuracy: {self.train_accuracies[-1]:.2f}%")
        print(f"  ├─ Final Train Loss: {self.train_losses[-1]:.4f}")
        print(f"  └─ Final Val Loss: {self.val_losses[-1]:.4f}")
        print(f"\nClass Distribution:")
        for i, class_name in enumerate(self.class_names):
            print(f"  {i:2d}. {class_name}")
        print(f"{'='*80}\n")


# ============================================================================
# MAIN TRAINING SCRIPT
# ============================================================================

if __name__ == "__main__":

    print("\n")
    print("╔" + "="*78 + "╗")
    print("║" + " "*78 + "║")
    print("║" + "  ADVANCED CNN TRAINING - MAXIMUM AUGMENTATION".center(78) + "║")
    print("║" + "  High Accuracy Fruit Classification".center(78) + "║")
    print("║" + " "*78 + "║")
    print("╚" + "="*78 + "╝")
    print("\n")

    # ========================================================================
    # CONFIGURATION - UPDATE THESE SETTINGS
    # ========================================================================

    # Dataset path (must contain train/ and val/ subdirectories)
    DATASET_PATH = "/content/drive/MyDrive/Fruit_Classification"

    # Model selection (recommended: resnet50, efficientnet_b3, or densenet121)
    # Options: 'custom', 'resnet18', 'resnet50', 'resnet101',
    #          'efficientnet_b0', 'efficientnet_b3',
    #          'vgg16', 'vgg19', 'densenet121'
    MODEL_TYPE = 'resnet50'

    # Training hyperparameters
    IMG_SIZE = 224           # Image size (224 or 384)
    BATCH_SIZE = 32          # Batch size (reduce if out of memory)
    NUM_EPOCHS = 100         # Maximum epochs (early stopping will stop earlier)
    LEARNING_RATE = 0.001    # Initial learning rate
    USE_PRETRAINED = True    # Use ImageNet pretrained weights (recommended)

    # Output paths
    MODEL_SAVE_PATH = "best_fruit_classifier.pth"
    PLOT_SAVE_PATH = "training_history.png"

    # ========================================================================
    # TRAINING PIPELINE
    # ========================================================================

    # Initialize trainer
    trainer = AdvancedCNNTrainer(
        dataset_path=DATASET_PATH,
        model_type=MODEL_TYPE,
        img_size=IMG_SIZE,
        batch_size=BATCH_SIZE,
        num_epochs=NUM_EPOCHS,
        learning_rate=LEARNING_RATE,
        use_pretrained=USE_PRETRAINED
    )

    # Create data loaders
    trainer.create_data_loaders()

    # Build model
    trainer.build_model()

    # Train the model
    trainer.train()

    # Save the model
    model_path = trainer.save_model(MODEL_SAVE_PATH)

    # Plot training history
    plot_path = trainer.plot_training_history(PLOT_SAVE_PATH)

    # Print summary
    trainer.print_summary()

    # ========================================================================
    # DOWNLOAD MODEL FILE (for Google Colab)
    # ========================================================================

    try:
        from google.colab import files
        print(f"{'='*80}")
        print("DOWNLOADING MODEL FILE")
        print(f"{'='*80}")
        print(f"Downloading: {model_path}")
        files.download(model_path)
        print(f"\nDownloading: {plot_path}")
        files.download(plot_path)
        print(f"\n✓ Downloads complete!")
        print(f"{'='*80}\n")
    except:
        print(f"{'='*80}")
        print("MODEL FILES SAVED LOCALLY")
        print(f"{'='*80}")
        print(f"Model saved to: {model_path}")
        print(f"Plot saved to: {plot_path}")
        print(f"\nIf using Colab, files are in your runtime.")
        print(f"If using local Jupyter, files are in your working directory.")
        print(f"{'='*80}\n")

    print("\n✓ TRAINING COMPLETE! Your model is ready to use.\n")
```
